### Assignment 10: Class with Properties

Create a class named `Rectangle` with private attributes `length` and `width`. Use properties to get and set these attributes. Create an object of the class and test the properties.

```python
class NegativeValueError(Exception):
    pass

class Rectangle:
    def __init__(self,length,width):
        self.__length=length
        self.__width=width
        
    @property                # this is  the length getter 
    def length(self):
        return self.__length
    
    @length.setter          # this is the  length setter 
    def length(self,length):
        if length<=0:
            raise NegativeValueError("Length can't be negative or zero")
        
        self.__length=length
        
    @property              # this is the width getter 
        
    def width(self):
        return self.__width
    
    @width.setter          # this is the width setter 
    def width(self,width):
        if width<=0:
            raise NegativeValueError("Width can't be negative or zero")
        
        self.__width=width 
          
 # Now making the object   
     
rect=Rectangle(10,5)
print(f"Before Updating:\nLength:{rect.length}\nWidth:{rect.width}")

# Now try to update the private length and width values
rect.length=20
rect.width=10
# now print the updated values
print(f"After Updating:\nLength:{rect.length}\nWidth:{rect.width}")
```

### Assignment 11: Abstract Base Class

Create an abstract base class named `Shape` with an abstract method `area`. Create derived classes `Circle` and `Square` that implement the `area` method. Create objects of the derived classes and call the `area` method.

```python
import math
from abc import ABC,abstractmethod
class Shape(ABC):  # ABC stands for Abstract Bas Class
    @abstractmethod
    def area(self):  # that tells that the class derived from that must have the area method init
        pass
    
class Circle(Shape):  # that is being inherited from the Shape Class
    def __init__(self,radius):
        self.radius=radius
        
    def area(self):
        return math.pi * self.radius**2

class Square(Shape):    # that is also being inherited from the Shape Class
    def __init__(self,side):
        self.side=side
        
    def area(self):
        return self.side**2
    
circle = Circle(10) # remember that we can't make the direct object of the base class Shape it will us error
square = Square(5)
print(f"Circle Area:{circle.area()}\n") 
print(f"Square Area:{square.area()}\n")  
```

### Assignment 12: Operator Overloading

Create a class named `Vector` with attributes `x` and `y`. Overload the `+` operator to add two `Vector` objects. Create objects of the class and test the operator overloading.

```python
class Vector:
    def __init__(self,x,y):
        self.x=x
        self.y=y
        
    def __add__(self,others):
        return Vector(self.x + others.x,self.y + others.y) # this will add two vectors corresponding valuess
    
    def __str__(self): # this will match the v3 format
        return f"Vector({self.x},{self.y})"
    
v1=Vector(2,4)
v2=Vector(4,8)
v3=v1 + v2  # sees operator overloading v1.__add__(v2)
print(v3)
```

### Assignment 14: Class with Context Manager

Create a class named `FileManager` that implements the context manager protocol to open and close a file. Use this class to read the contents of a file.

```python
class FileManager:
    def __init__(self, filename, mode):
        self.filename = filename
        self.mode = mode

    def __enter__(self):
        self.file = open(self.filename, self.mode)
        return self.file

    def __exit__(self, exc_type, exc_value, traceback):
        self.file.close()


with FileManager('sample.txt', 'r') as file:
    content = file.read()
    print(content)
```

### Assignment 15: Chaining Methods

Create a class named `Calculator` with methods to add, subtract, multiply, and divide. Each method should return the object itself to allow method chaining. Create an object and chain multiple method calls.

```python
class Calculator:
    def __init__(self, value=0):
        self.value = value

    def add(self, amount):
        self.value += amount
        return self

    def subtract(self, amount):
        self.value -= amount
        return self

    def multiply(self, amount):
        self.value *= amount
        return self

    def divide(self, amount):
        if amount != 0:
            self.value /= amount
        else:
            print("Cannot divide by zero!")
        return self


calc = Calculator()
calc.add(10).subtract(3).multiply(2).divide(2)
print(calc.value)  # 7.0
```
